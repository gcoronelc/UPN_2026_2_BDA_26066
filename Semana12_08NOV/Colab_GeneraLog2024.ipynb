```python
from datetime import datetime
from faker import Faker
import random

# Define la ruta y el nombre del archivo de registro en Google Drive
log_file_path = '/content/drive/MyDrive/logs/log_file.txt'

# Redirige la salida estándar al archivo de registro
import sys
sys.stdout = open(log_file_path, 'w')

# Crea una instancia de Faker para generar datos ficticios
fake = Faker()

# Genera datos ficticios y escribe en el archivo de registro
for _ in range(10000):
    # Genera la fecha y hora actual
    now = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    # Genera un éxito o error aleatorio
    success = random.choice([True, False])

    # Genera una dirección IP ficticia
    ip_address = fake.ipv4()

    # Nombre Ficticio
    name_person = fake.name()

    # Correo Ficticio
    mail_person = fake.email()

    # Pais Ficticio
    country_person = fake.country()

    #tiempo de espera
    espera_ms_success = random.randint(5, 140)
    espera_ms_error = random.randint(150, 400)

    # Escribe los datos en el archivo de registro
    if success:
        print(f"{now},SUCCESS,{ip_address},{name_person} {mail_person},{country_person},{espera_ms_success}")
    else:
        print(f"{now},ERROR,{ip_address},{name_person} {mail_person},{country_person},{espera_ms_error}")

# Cierra el archivo y restaura la salida estándar original
sys.stdout.close()
sys.stdout = sys.__stdout__
```
